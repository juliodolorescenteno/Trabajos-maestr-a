El algoritmo Johnson(1986) genera variables aleatorias $(u,t)$ desde una distribución $FGM$ con parámetro $\theta$:
1. Generar variables $(u,t)$ uniformes independientes $(0,1)$
2. Sea $a=1+\theta(1-2u)$ y \
  $b=\sqrt(a^2-4(a-1)t)$
3. Sea $v=\frac{2t}{(b+a)}$
4. Asignar pares $(u,v)$

```python
import numpy as np
import matplotlib.pyplot as plt

def generar_copula_FGM(n, theta):

    ###Generamos las variables aleatorias
    u = np.random.uniform(0, 1, n)
    t = np.random.uniform(0, 1, n)


    a=1+theta*(1-2*u)
    b=np.sqrt(a**2-4*(a-1)*t)
    v=(2*t)/(b+a)

    return u, v


casos = [1, -0.5, 0.5, -1] #Casos donde θ=[1,-1] (ejemplo dado en el libro)
colores = ['purple','blue','royalblue','blueviolet']
fig, axes = plt.subplots(2, 2, figsize=(10, 10))
for ax, theta, color in zip(axes.ravel(), casos, colores):
    u, v = generar_copula_FGM(5000, theta)
    ax.scatter(u, v, alpha=0.35, s=2, color=color)
    ax.set_title(f'Cópula FGM\nθ={theta}', fontsize=12)
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_xlabel('U')
    ax.set_ylabel('V')
    ax.grid(True, alpha=0.1)
plt.tight_layout()
plt.show()
```

```python
### Mapas de calor
fig, axes = plt.subplots(2, 2, figsize=(10, 10))
for ax, theta in zip(axes.ravel(), casos):
    u, v = generar_copula_FGM(5000, theta)
    hb = ax.hist2d(u, v, bins=40, range=[[0, 1], [0, 1]], cmap="viridis")
    ax.set_title(f'Cópula FGM\nθ={theta}', fontsize=12)
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_xlabel('U')
    ax.set_ylabel('V')
    plt.colorbar(hb[3], ax=ax)

plt.tight_layout()
plt.show()
```
